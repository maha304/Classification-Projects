```python
# Re-import required packages and reload the modified DataFrame structure
import pandas as pd

# Reload original dataset
df = pd.read_csv('/mnt/data/noshowappointments-2016.csv', encoding='latin1')

# Clean column names and convert values
df.rename(columns={
    'Hipertension': 'Hypertension',
    'Handcap': 'Handicap',
    'No-show': 'No_show'
}, inplace=True)

# Convert binary values to Yes/No
binary_map = {0: 'No', 1: 'Yes'}
binary_cols = ['Scholarship', 'Hypertension', 'Diabetes', 'Alcoholism', 'SMS_received']
for col in binary_cols:
    if col in df.columns:
        df[col] = df[col].map(binary_map)

# Map No_show values
df['No_show'] = df['No_show'].map({'No': 'Showed', 'Yes': 'No-show'})

# Convert date columns and create new features
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay']).dt.date
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay']).dt.date
df['Days_between'] = (pd.to_datetime(df['AppointmentDay']) - pd.to_datetime(df['ScheduledDay'])).dt.days
df = df[df['Days_between'] >= 0]
df['DayOfWeek'] = pd.to_datetime(df['AppointmentDay']).dt.day_name()
df['ScheduledHour'] = pd.to_datetime(df['ScheduledDay']).dt.hour

# Save cleaned data
cleaned_data_path = "/mnt/data/cleaned_noshow_data.csv"
df.to_csv(cleaned_data_path, index=False)

cleaned_data_path
```
